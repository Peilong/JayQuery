## Binary Label Encoding
So given the distribution and my research, pass & distinction should definitely be considered a success on the final label. Fail is obviously a non-success, but withdrawn is tricky. Theoretically it is not a Fail or Pass, but it is still valuable data, even with the ambigious meaning. I think in the end, it is best to put withdrawn as a non-success. While it isn't as destructive as a Fail, it slows the progress of an individual through their courses, and impedes the expectation of a timely graduation. 

```python
dfs['studentInfo']
```

```python
sidf = dfs['studentInfo'].drop(labels=['final_result'],axis=1)
sidf.dropna(subset=['imd_band'], inplace=True)
```

```python
sidf['success'] = dfs['studentInfo']['final_result'].map({'Pass':1,
                                                   'Distinction':1,
                                                   'Fail':0,
                                                   'Withdrawn':0})
```

## Silo A Features
imd_band (deprivation index, ordinal-encoded), highest_education (ordinal), age_band (ordinal), gender (binary), disability (binary), region (one-hot, 8 dims), studied_credits (continuous, normalized)

```python
sidf.columns
```

```python
a_df = sidf.drop(axis=1, labels=['code_module','code_presentation','id_student'])
# Silo A includes region, which I think should be somewhat pointless for our uses (as we wont be deploying with UK school data)
# Silo A also removes num_of_prev_attempts, which I think could still have some use
# I ended up leaving both region and num_of_prev_attempts in, as I can remove them later if they are not showing much use.
```

### Ordinal Encoding

```python
a_df.columns
```

```python
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import MinMaxScaler
```

```python
enc = OrdinalEncoder()
a_df_e = pd.DataFrame(a_df['success'])
a_df_e['imd_band'] = enc.fit_transform(a_df[['imd_band']])
#encoded 0-10% as 0, and 90-100% as 9
```

```python
sorted(a_df['highest_education'].unique())
```

```python
a_df_e['highest_education'] = a_df['highest_education'].map({
    'No Formal quals':0,
    'Lower Than A Level':1,                                                            
    'A Level or Equivalent':2,
    'HE Qualification':3,
    'Post Graduate Qualification':4})
```

```python
sorted(a_df['age_band'].unique())
```

```python
a_df_e['age_band'] = enc.fit_transform(a_df[['age_band']])
```

### Binary Encoding

```python
sorted(a_df['gender'].unique())
```

```python
a_df_e['gender'] = a_df['gender'].map({'M': 1, 'F': 0})
```

```python
sorted(a_df['disability'].unique())
```

```python
a_df_e['disability'] = a_df['disability'].map({'Y':1,'N':0})
```

### One Hot Encoding

```python
ohe = OneHotEncoder(drop='first', sparse_output=False)
```

```python
region_encoded = ohe.fit_transform(a_df[['region']])
region_encoded = pd.DataFrame(
    region_encoded, 
    columns=ohe.get_feature_names_out(['region'])
)
```

### Normalization

```python
scaler = MinMaxScaler()
a_df_e['studied_credits'] = scaler.fit_transform(a_df[['studied_credits']])
```

```python
silo_a_encoded_df = pd.concat([a_df_e.reset_index(drop=True),region_encoded.reset_index(drop=True)],axis=1)
```
