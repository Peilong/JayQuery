## Binary Label Encoding
So given the distribution and my research, pass & distinction should definitely be considered a success on the final label. Fail is obviously a non-success, but withdrawn is tricky. Theoretically it is not a Fail or Pass, but it is still valuable data, even with the ambigious meaning. I think in the end, it is best to put withdrawn as a non-success. While it isn't as destructive as a Fail, it slows the progress of an individual through their courses, and impedes the expectation of a timely graduation.

```python
silo_b_df = pd.merge(
    dfs['studentVle'],
    dfs['vle'].drop(axis=1,columns=['code_module','code_presentation','week_from','week_to']), 
    on='id_site',how='left').drop(axis=1,columns=['code_module','code_presentation'])
```

```python
silo_b_df
```

```python
silo_b_agg = (
    silo_b_df.groupby(['id_student', 'id_site', 'date', 'activity_type'], as_index=False)['sum_click']
    .sum()
)
```

```python
silo_b_agg
```

```python
silo_b_agg_date_reduced = (
    silo_b_agg.groupby(['id_student', 'id_site','activity_type'], as_index=False)
    .agg(
        start_date=('date', 'min'),
        end_date=('date', 'max'),
        active_days=('date', 'nunique'),
        sum_click=('sum_click','sum')
    )
)

silo_b_agg_date_reduced
```

```python
activity_df = pd.get_dummies(dfs['vle'],columns=['activity_type'], prefix='vle_clicks').drop(axis=1,columns=['week_from','week_to'])
silo_b_agg_enc = pd.merge(silo_b_agg_date_reduced, activity_df, on='id_site', how='left').drop(axis=1,labels=['activity_type'])
silo_b_agg_enc = silo_b_agg_enc[silo_b_agg_enc['code_module']=='CCC']
silo_b_agg_enc = silo_b_agg_enc[silo_b_agg_enc['code_presentation']=='2014J']
silo_b_agg_enc.drop(axis=1,columns=['code_module','code_presentation'],inplace=True)
silo_b_agg_enc
```

```python
vle_cols = [col for col in silo_b_agg_enc.columns if col.startswith('vle_clicks_')]


agg_dict = {
    'start_date': 'min',
    'end_date': 'max',
    'active_days': 'sum',
    'sum_click': 'sum',
    **{col: 'any' for col in vle_cols}
}

silo_b_vle = silo_b_agg_enc.groupby('id_student', as_index=False).agg(agg_dict)
silo_b_vle
```

```python
empty_cols = [col for col in vle_cols if not silo_b_vle[col].any()]
print(empty_cols)
```

```python
m_code = 'CCC'
p_code = '2014J'

# Get the full list of activity types across ALL modules
all_activity_types = dfs['vle']['activity_type'].unique()
all_vle_cols = [f'vle_clicks_{a}' for a in all_activity_types]

merged = (
    dfs['studentVle']
    .merge(dfs['vle'].drop(columns=['week_from', 'week_to', 'code_module','code_presentation']), on='id_site', how='left')
    .query("code_module == @m_code and code_presentation == @p_code")
    .drop(columns=['code_module', 'code_presentation'])
)

activity_clicks = (
    merged
    .groupby(['id_student', 'activity_type'], as_index=False)
    .agg(sum_click=('sum_click', 'sum'))
    .pivot_table(index='id_student', columns='activity_type', values='sum_click', fill_value=0)
    .rename(columns=lambda x: f'vle_clicks_{x}')
    .reindex(columns=all_vle_cols, fill_value=0)  # enforce all columns
    .reset_index()
)

student_stats = (
    merged
    .groupby('id_student', as_index=False)
    .agg(
        start_date=('date', 'min'),
        end_date=('date', 'max'),
        active_days=('date', 'nunique')
    )
)

vle_df = student_stats.merge(activity_clicks, on='id_student', how='left')
vle_df
```

```python
empty_cols = [col for col in vle_cols if not silo_b_vle[col].any()]
print(empty_cols)
```

```python
assessment_df_score = (
    assessment_df.assign(weighted_score=(assessment_df['score'] * assessment_df['weight'])/100)
    .groupby('id_student', as_index=False)
    .agg(
        weighted_score=('weighted_score', 'sum'),
        total_weight = ('weight','sum'),
        num_assessments = ('id_assessment','nunique')
        )
)
assessment_df_score
```

```python
silo_b = pd.merge(assessment_df_score, silo_b_vle, on='id_student', how='left')
silo_b
```

```python
assessment_df = (
    dfs['studentAssessment']
    .merge(dfs['assessments'], on='id_assessment', how='left')
    .query("code_module == @m_code and code_presentation == @p_code")
    .drop(columns=['is_banked', 'code_module', 'code_presentation', 'date'])
    .assign(weighted_score=lambda x: (x['score'] * x['weight']) / 100)
    .groupby('id_student', as_index=False)
    .agg(
        weighted_score=('weighted_score', 'sum'),
        total_weight=('weight', 'sum'),
        num_assessments=('id_assessment', 'nunique')
    ).assign(
        participation_rate=lambda x: x['total_weight'] / 200,
        weighted_score_pct=lambda x: x['weighted_score'] / 200
    ).drop(columns=['total_weight','weighted_score'])
)
    
silo_b = assessment_df.merge(vle_df, on='id_student', how='left')
silo_b
```
